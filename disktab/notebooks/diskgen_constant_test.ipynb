```python
import numpy as np
from astropy import units as u
from astropy.constants import c
import sys, os
sys.path.append(os.path.abspath('..'))
from disktab import AGNDisk
```

```python
M = 1e8 * u.Msun
lambda_d = 0.01
eta_d = 0.1
alpha = 0.1

L_edd = 1.26e38 * (M / u.Msun) * u.erg / u.s
Mdot = lambda_d * L_edd / (eta_d * c**2)

disk = AGNDisk()
disk.M = M
disk.alpha = alpha
disk.Mdot = Mdot.to(u.kg / u.s)

rvals = np.logspace(1, 9, 5000)
df = disk.generate_constant_test_model(rvals=rvals, zone=1)
df
```
