```python
import sys, os
sys.path.insert(0, os.path.abspath('../..'))  # for spacehub_calc
import spacehub_calc as shc
import matplotlib.pyplot as plt
import numpy as n
from IPython.display import HTML
```

```python
orb = shc.TwoBodyOrbit("../../SpaceHub/test/nick_test/disk_model/AnalyticalTests/DiskModel-Circular.dat", mute=True)
print(orb)
```

```python
fig, axs = orb.plot_keplerian_evolution(plots=('a', 'e'), ylim=False)
axs[0].set_xlabel('Time (years / 2π)')
axs[1].set_xlabel('Time (years / 2π)')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(orb.years, orb.semiMajorAxis, label='Semi-major axis')
ax.set_xlabel('Time (years)')
ax.set_ylabel('a (AU)')
ax.set_title('Orbital Decay from Disk Drag')
ax.legend()
plt.show()
```

```python
fig, axs = orb.plot_orbit_3panel()
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ani = orb.plot_trajectory_3d(fig, ax, frames=500, interval=20)
HTML(ani.to_jshtml())
```

```python
a0 = orb.semiMajorAxis[0]
a_final = orb.semiMajorAxis[-1]
delta_a = a_final - a0
t_final = orb.years[-1]

print(f"Initial a: {a0:.4f} AU")
print(f"Final a: {a_final:.4f} AU")
print(f"Change in a: {delta_a:.4f} AU ({100*delta_a/a0:.2f}%)")
print(f"Simulation time: {t_final:.2f} years")
```
