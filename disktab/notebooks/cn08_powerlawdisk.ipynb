```python
from astropy import units as u
from astropy.constants import G, c, M_jup
import numpy as np
from numpy import pi
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
import sys
import os

#references
#CN08: 2008A&A...482..677C
```

```python
R0 = 0.1* u.AU 
Rf = 40*u.AU
Rvals = [r * u.AU for r in np.logspace(-1, 1.6020599913279625, 5000)]
Sigma0 = 22 * 40 * M_jup * (1.5/((Rf)**(1.5)-(R0**(1.5)))) / (2*pi) # factor of 22 added to get the curves to line up with CN08 Figures 1-3

Hs = []
Sigmas = []
for R in Rvals:
    SigmaR = Sigma0 * R**(-0.5)
    Sigmas.append(SigmaR.to(u.Msun / u.AU / u.AU))
    H = 0.05*R 
    Hs.append(H.to(u.AU))
```

```python
df = pd.DataFrame({
    'R': [R.to(u.AU).value for R in Rvals],
    'R/Rg' : [-1 for R in Rvals],
    'Tc': [-1 for R in Rvals],
    'rho': [-1 for R in Rvals],
    'P': [-1 for R in Rvals],
    'cs' : [-1 for R in Rvals],
    'H': [H.to(u.AU).value for H in Hs],
    'visc': [-1 for R in Rvals],
    'Sigma': [S.value for S in Sigmas],
    'Q': [-1 for R in Rvals],
    'grad_T': [-1 for R in Rvals],
    'grad_Sigma': [0.5 for R in Rvals],
    'grad_P': [-1 for R in Rvals],
    'gamma':[-1 for R in Rvals],
    'f_thermal':[-1 for R in Rvals],
    'v_disk':[-1 for R in Rvals]
})

df.to_csv("../../SpaceHub/src/interaction/disk_tab/cn08_powerlaw_3col.csv", index=None)
```
