```python
import sys, os
sys.path.insert(0, os.path.abspath('../..'))
import spacehub_calc as shc
import matplotlib.pyplot as plt
from IPython.display import HTML
```

## Interior to Q=1 (1000 AU, retrograde)

```python
orb_retro = shc.TwoBodyOrbit('../../SpaceHub/test/nick_test/disk_model/AnalyticalTests/DiskModel-Retrograde.dat', mute=True)
print(orb_retro)
```

```python
fig, axs = orb_retro.plot_keplerian_evolution(plots=('a', 'e', 'i', 'omega'), ylim=False)
plt.suptitle('Interior to Q=1 (1000 AU) - Keplerian Evolution')
plt.tight_layout()
plt.show()
```

```python
fig, axs = orb_retro.plot_orbit_3panel()
plt.suptitle('Interior to Q=1 (1000 AU) - Trajectory')
plt.tight_layout()
plt.show()
```
