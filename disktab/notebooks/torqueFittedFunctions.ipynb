```python
from astropy import units as u
from astropy.constants import G,c
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt 
```

```python

Sigmap = 1e-10 *u.g/u.cm/u.cm 
a = 100*u.AU 
r = 100*u.AU
hp = 100*u.km / r 
Mp = 1*u.Mearth
Mstar = 1 *u.Msun
Omegap = np.sqrt(G*Mstar/a)

FJ0 = Sigmap * a**4 * Omegap**2 * hp**(-3) * (Mp/Mstar)**2

def Tnet(p,q,hp):
    return (1.413+1.109*q+1.237*p-0.017*q*p)*hp**(1.025)*FJ0

```
