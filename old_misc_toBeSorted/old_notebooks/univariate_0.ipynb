```python
from astropy import units as u
from astropy.table import Table
from astropy.constants import G, c, k_B, sigma_sb, m_p
mu = 0.6
import simple_disk as sd
from matplotlib import pyplot as plt
import disk_newton_solver as newton
import numpy as np
from numpy import pi as pi
from disk_newton_solver import Tester
import pandas as pd
import seaborn as sns
```

```python
#Free vars
M = 1e8 * u.Msun
Mdot = 1e23 * u.kg / u.s 
alpha = 0.1
Rstar = 6 * (G * M / c / c)

Tc = 1e6 * u.K
k0 = 0.4 * u.cm**2 / u.g
k1 = 6.4e22 * u.cm**5 / u.g**2 * u.K**(7/2)
```

```python
R = 10 * Rstar

f = (1-np.sqrt(Rstar/R))
Omega = np.sqrt(G*M / R**3)

A = k_B * Tc / (mu * m_p)
B = 4 * sigma_sb * Tc**4 / (3*c) 
C = (Mdot * f * Omega ** 2 / (3 * pi * alpha) ) ** (2/3)

D = k0
E = k1 * Tc**(-7/2)
F = (32 * pi * sigma_sb * Omega * R**3) * (3 * pi * alpha) * Tc**4 / (9 * G * M * Mdot * f) / (Mdot * f * Omega**2)**(1/3)
a = C * E / F
b = (C * D / F) - A 
cee = - B 
rho_plus = (-b + np.sqrt(b**2 - 4*a*cee))/(2*a)
rho_minus = (-b - np.sqrt(b**2 - 4*a*cee))/(2*a)
    
```
