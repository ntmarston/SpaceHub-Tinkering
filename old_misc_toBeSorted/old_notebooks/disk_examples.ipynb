# Shakura-Sunyaev Disk Properties
## Examples using Table 3.1 functions

````python
from astropy import units as u
from simple_disk import *
````

## Example 1: AGN Disk (Zone 2)

````python
# AGN parameters
alpha = 0.1
M = 1e8 * u.Msun
fMdot = 1 * u.Msun / u.yr  # Accretion rate
r = 10 * (G * M / c**2)  # 10 gravitational radii
zone = 2
scale = "AGN"

print("=" * 50)
print(f"AGN Disk Properties (Zone {zone})")
print("=" * 50)
print(f"Black hole mass: {M}")
print(f"Accretion rate: {fMdot.to(u.Msun/u.yr)}")
print(f"Radius: {r.to(u.AU):.2e}")
print(f"Alpha: {alpha}")
print("\nDisk Properties:")
print("-" * 50)
print(f"Disk height h: {h(alpha, M, fMdot, r, zone, scale).to(u.AU):.3e}")
print(f"Kepler frequency Ω: {Omega(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Period: {period(alpha, M, fMdot, r, zone, scale).to(u.day):.3e}")
print(f"Radial velocity vr: {vr(alpha, M, fMdot, r, zone, scale).to(u.km/u.s):.3e}")
print(f"Mass density ρ: {rho(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Surface density Σ: {Sigma(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Total pressure p: {pressure(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Pressure ratio (pgas/prad): {pressure_ratio(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Central temperature T: {T_central(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Surface temperature Ts: {T_surface(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Surface flux Q: {Q(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Optical depth (scattering) τes: {tau_es(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"True optical depth τ*: {tau_star(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Optical depth ratio τff/τes: {tau_ratio(alpha, M, fMdot, r, zone, scale):.3e}")
print(f"Maximum magnetic field Bmax: {B_max(alpha, M, fMdot, r, zone, scale):.3e}")
````

## Example 3: Compare All Three Zones
### AGN at 10 gravitational radii

````python
# AGN parameters
alpha = 0.1
M = 1e8 * u.Msun
fMdot = 1 * u.Msun / u.yr
r = 10 * (G * M / c**2)
scale = "AGN"

print("=" * 70)
print(f"Comparison of Zones at r = 10 GM/c² (AGN)")
print("=" * 70)
print(f"{'':<30} {'Zone 1':<15} {'Zone 2':<15} {'Zone 3':<15}")
print("-" * 70)

for zone in [1, 2, 3]:
    if zone == 1:
        h1 = h(alpha, M, fMdot, r, zone, scale).to(u.AU)
        T1 = T_central(alpha, M, fMdot, r, zone, scale)
        rho1 = rho(alpha, M, fMdot, r, zone, scale)
    elif zone == 2:
        h2 = h(alpha, M, fMdot, r, zone, scale).to(u.AU)
        T2 = T_central(alpha, M, fMdot, r, zone, scale)
        rho2 = rho(alpha, M, fMdot, r, zone, scale)
    else:
        h3 = h(alpha, M, fMdot, r, zone, scale).to(u.AU)
        T3 = T_central(alpha, M, fMdot, r, zone, scale)
        rho3 = rho(alpha, M, fMdot, r, zone, scale)

print(f"{'Disk height h [AU]':<30} {h1.value:13.3e}   {h2.value:13.3e}   {h3.value:13.3e}")
print(f"{'Central temp T [K]':<30} {T1.value:13.3e}   {T2.value:13.3e}   {T3.value:13.3e}")
print(f"{'Mass density ρ [kg/m³]':<30} {rho1.value:13.3e}   {rho2.value:13.3e}   {rho3.value:13.3e}")
````

## Example 4: Radial Profile
### Plot disk height vs radius for zone 2

````python
import numpy as np
import matplotlib.pyplot as plt

# AGN parameters
alpha = 0.1
M = 1e8 * u.Msun
fMdot = 1 * u.Msun / u.yr
zone = 2
scale = "AGN"

# Create array of radii from 6 to 100 GM/c²
r_grav = np.logspace(np.log10(6), 2, 50)  # in units of GM/c²
r_array = r_grav * (G * M / c**2)

# Calculate disk height for each radius
h_array = [h(alpha, M, fMdot, r_val, zone, scale).to(u.AU).value for r_val in r_array]
T_array = [T_central(alpha, M, fMdot, r_val, zone, scale).value for r_val in r_array]

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Disk height
ax1.loglog(r_grav, h_array, 'b-', linewidth=2)
ax1.set_xlabel('Radius [GM/c²]', fontsize=12)
ax1.set_ylabel('Disk Height h [AU]', fontsize=12)
ax1.set_title(f'AGN Disk Height (Zone {zone}, α={alpha})', fontsize=14)
ax1.grid(True, alpha=0.3)

# Central temperature
ax2.loglog(r_grav, T_array, 'r-', linewidth=2)
ax2.set_xlabel('Radius [GM/c²]', fontsize=12)
ax2.set_ylabel('Central Temperature T [K]', fontsize=12)
ax2.set_title(f'AGN Disk Temperature (Zone {zone}, α={alpha})', fontsize=14)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Quick Reference

All functions have the same signature:
```python
property(alpha, M, fMdot, r, zone, scale="AGN")
```

Where:
- `alpha`: Viscosity parameter (dimensionless)
- `M`: Black hole mass (with astropy units, e.g., `1e8 * u.Msun`)
- `fMdot`: Accretion rate (with astropy units, e.g., `1 * u.Msun/u.yr`)
- `r`: Radius (with astropy units, e.g., `10 * (G*M/c**2)` or `1 * u.AU`)
- `zone`: Disk zone (1, 2, or 3)
- `scale`: "AGN" or "galactic"

Available functions:
1. `h()` - Disk half thickness
2. `Omega()` - Kepler frequency
3. `period()` - Orbital period
4. `radial_time()` - Radial time r/vr
5. `vr()` - Radial velocity
6. `rho()` - Mass density
7. `Sigma()` - Surface density
8. `pressure()` - Total pressure
9. `pressure_ratio()` - pgas/prad
10. `T_central()` - Central temperature
11. `T_surface()` - Surface temperature
12. `Q()` - Surface flux
13. `tau_es()` - Optical depth (scattering)
14. `tau_star()` - True optical depth
15. `tau_ratio()` - τff/τes
16. `B_max()` - Maximum magnetic field
